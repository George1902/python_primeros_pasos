#**2. Datos Númericos**

```python
## Separando las columnas del DataFrame para mejor tratamiento de los datos.

columnas = list(datos.columns)
columnas
```

```python
## Utilizamos el método explode para visualizar de manera mas detallada desde la columna 3 en adelante, creando un nuevo registro.
## Ayuda a eliminar las listas del DataFrame

datos = datos.explode(columnas[3:])
datos.head()
```

```python
## Reseteo de indice, luego de aplicar el método explode el indice queda en 0 para todos lo items, se debe reinicar de la siguiente manera:

datos = datos.reset_index(drop=True)
datos.head()
```

```python
## Examinar nuestro nuevo DataFrame

datos.info()
```

```python
## Para tratar los object del DataFrame(Son caracteres indeterminados por Phyton) importamos la biblioteca Numpy

import numpy as np
```

```python
## A la columna (max_hospedes) le cambiamos el tipo de datos de object a entero

datos['max_hospedes'] = datos['max_hospedes'].astype(np.int64)
```

```python
datos.info()
```

```python
## Cambiando con una lista el tipo de entrada de los datos

col_numericas = ['cantidad_baños', 'cantidad_cuartos', 'cantidad_camas' ]
```

```python
datos[col_numericas] = datos[col_numericas].astype(np.int64)
```

```python
datos.info()
```

```python
## A la columna (evaluacion_general) le cambiamos el tipo de datos de object a decimal

datos['evaluacion_general'] = datos['evaluacion_general'].astype(np.float64)
```

```python
datos.info()
```

```python
## Se debe cambiar el tipo de dato de las columnas que tienen valores de texto y númerico ($100) a solo númerico.

datos['precio']
```

```python
## Aplicando el método apply mas la función lambda se logra una transformación en cada uno de los valores de la columna.
## Con lambda la X se convierte en cada items del DataFrame y recorre todos los datos mas el replace para cambiar los caracteres.
## Strip quita los espacios adelante, atras luego de haber retirado los caracteres anteriores.

datos['precio'] = datos['precio'].apply(lambda x: x.replace('$', '').replace(',', '').strip())
```

```python
datos['precio']
```

```python
datos['precio'] = datos['precio'].astype(np.float64)
```

```python
datos.info()
```

```python
## Para Inspeccionar dos o mas columnas se debe colocar siempre doble corchete.

datos[['cuota_deposito','cuota_limpieza']]
```

```python
## Eliminando los caracteres no deseados en multiples columnas.
## Cuando trabajamos con una columna es (Series).
## Cuando trabajamos con mas de una columna es (DataFrame).
## El .apply solo funciona con series unicamente con una columna.
## Applymap funciona para dos o mas columnas.

datos[['cuota_deposito','cuota_limpieza']] = datos[['cuota_deposito','cuota_limpieza']].applymap(lambda x: x.replace('$', '').replace(',', '').strip())
```

```python
datos[['cuota_deposito','cuota_limpieza']] = datos[['cuota_deposito','cuota_limpieza']].astype(np.float64)
```

```python
datos.info()
```

#**3. Datos de Texto**

```python
datos['descripcion_local'] = datos['descripcion_local'].str.replace('\n', ' ')
```

```python
datos.head()
```

```python
datos['descripcion_local']
```

```python
from typing import assert_type
## Para tratar los datos como token primero se debe llevarlos a minuscula.
## Transformamos la columna a un string.
## Método lower() devuelve o retorna una copia en minuscula.

datos['descripcion_local'] = datos['descripcion_local'].astype(str).str.lower()
```

```python
datos.head()
```

```python
## Examinando como el ejemplo el indice 3169.

datos['descripcion_local'][3169]
```

```python
## Método regex=True para hcaer un reemplazo de caracteres no deseados mas complejo.
## ['^a-zA-Z0-9\-\'] Signifa que va a reemplazar todo lo contrario a esos caracteres, sustituye todo menos las letras y los números.

# Fixed SyntaxWarning: Used a raw string for the regex pattern to correctly handle backslashes.
datos['descripcion_local'] = datos['descripcion_local'].str.replace(r"[^a-zA-Z0-9' -]", ' ', regex=True)
```

```python
## '(?<!\w)-(?<!\w)' Reemplaza los guiones, busca los guiones en palabras que no tengan palabras o caracteres acompañados a la izquierda y derecha.

datos['descripcion_local'] = datos['descripcion_local'].str.replace('(?<!\\w)-(?<!\\w)', ' ',regex=True)
```

```python
datos['descripcion_local'].head()
```

```python
## El split devuelve una lista seperando cada espacio (token) como una palabra independiente.

datos['descripcion_local'] = datos['descripcion_local'].astype(str).str.split()
datos.head()
```

```python
## Limpiar un poco la columna (comodidades).
## '\{|}|' La barra indica buscar un caracter especifico, en este caso la llave de entrada de la columna comodidades.
## La barra horizontal | se utiliza como separador para que busque otro elemento, en este caso la llave (}) de cierre.
## Luego otra barra horizontal | para separar y buscar las comillas dobles (").

datos['comodidades'].str.replace('\{|}|\"','', regex=True)
```

```python
datos['comodidades'] = datos['comodidades'].str.replace('\{|}|\"','', regex=True)
```

```python
datos.head()
```

```python
## Construir una lista a la columna (comodidades)
## En este caso el split hay que indicar  que el separador de los token es una (,) que viene por defecto en la columna.

datos['comodidades'] = datos['comodidades'].str.split(',')
datos.head()
```

```python
dt_data = pd.read_json('/content/drive/MyDrive/Colab Notebooks/inmuebles_disponibles.json')
dt_data.head()
```

```python
## Cambiar el tipo de entrada de la columna ('fecha')

dt_data['fecha'] = pd.to_datetime(dt_data['fecha'])
```

```python
dt_data.info()
```

```python
dt_data.head()
```

```python
## Agruparemos las fechas para trabajar por mes, saber cuantas viviendas estuvieron disponibles por mes.
## strftime Personaliza la variable de fecha en la froma que querramos.
## (%Y) Indicamos que muestre el año.
## (%m) Indicamos que muestre el mes.

dt_data['fecha'].dt.strftime('%Y-%m')
```

```python
## Agrupamos para saber cuantas viviendas estaban disponibles por mes.
## Lo enlazamos con la columna (lugar_disponible) para saber la disponibilidad de las viviendas.
## ['lugar_disponible'].sum() Utilizando el método .sum() en una columna de booleanos solo indentica los valores True.

dt_data.groupby(dt_data['fecha'].dt.strftime('%Y-%m'))['lugar_disponible'].sum()
```
