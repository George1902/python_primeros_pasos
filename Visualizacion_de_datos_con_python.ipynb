# Visualización de datos con Python: Matplotlib, Seaborn y Plotly

```python
## Importamos la Biblioteca de Pandas.
## Almacenamos en una variable el Data Frame.

import pandas as pd

df = pd.read_csv('/content/drive/MyDrive/Colab Notebooks/inmigrantes_canada.csv')
df.head()
```

```python
## Con .shape observamos cuantos registros tiene el arichvo.

df.shape
```

```python
##  El .info nos ayuda a observr el tipo de dato que tenemos.

df.info()
```

```python
## El set_index ayuda a configurar el indice en el DataFrame, en este caso utilizar como indice el Pais.

df.set_index('Pais',inplace=True)
df
```

```python
## Creamos una variable de los años con el método list para poder mapear con str (strings) un rango que vaya desde 1980 hasta 2014

anos = list(map(str,range(1980,2014)))
anos
```

```python
## Aplicamos el filtro de años con el pais que deseamos obtener la cantidad de migrantes correspondientesa a cada año en este caso de Colombia.
## El .loc nos ayuda a encontrar el pais 'Colombia' dentro de este filtro que acabamos de crear con la variable 'anos'

colombia = df.loc['Colombia',anos]
colombia
```

```python
## Creamos un diccionario con el 'Año' el objeto colombia tomando los indices y pasarlo a una lista index.tolist.
## Segundo elemento es la cantidad de inmigrantes de colombia los valores (values).

col_dict = {'Año':colombia.index.tolist(),
            'Inmigrantes':colombia.values.tolist()}
col_dict
```

```python
## Creamos un DataFrame llamado datos_col y observamos con el .tail los últimos registros de la tabla.

datos_col = pd.DataFrame(col_dict)
datos_col.tail()
```

## Matplotlib
https://matplotlib.org/

```python
## Importamos la Biblioteca Matplotlib.

import matplotlib.pyplot as plt
```

```python
## Utilizamos una gráfica con años en el eje X y cantidad de migrantes en el eje Y.
## plt.xticks(['1980','1985','1990','1995','2000','2005','2010']), xticks indicamos que la tabla tenga tendencia a mostrar la grafica cada 5 años.

plt.figure(figsize=(8,4))
plt.plot(datos_col['Año'],datos_col['Inmigrantes'])
plt.xticks(['1980','1985','1990','1995','2000','2005','2010'])
plt.show()
```

```python
## Mejoramo nuestro gráfico añadiendo el titulo: plt.title('Inmigración de colombianos hacia Canadá')
## Deescripción para eje X: plt.xlabel('Año')
## Descripción para eje Y: plt.xlabel('Número de Inmigrantes')
## Importante el plt.show()  debe ser siempre el final del codigo.

plt.figure(figsize=(8,4))
plt.plot(datos_col['Año'],datos_col['Inmigrantes'])
plt.xticks(['1980','1985','1990','1995','2000','2005','2010'])
plt.title('Inmigración de Colombianos hacia Canadá')
plt.xlabel('Año')
plt.ylabel('Número de Inmigrantes')
plt.show()
```

```python
## Generamos las variables que nos van a permitir sostener el lienzo y la grafica de los datos.
## Con ax.xaxis.set_major_locator(plt.MultipleLocator(5)) agiliza el intervalo de los años de 5 en 5.

fig, ax = plt.subplots(figsize=(8,4))
ax.plot(datos_col['Año'],datos_col['Inmigrantes'])
ax.xaxis.set_major_locator(plt.MultipleLocator(5))
ax.set_title('Inmigración de Colombianos hacia Canadá\n1980 - 2013')
ax.set_xlabel('Año')
ax.set_ylabel('Número de Inmigrantes')
plt.show()
```

```python
## Gráfico de caja.
## Con plt.subplots(1,2,figsize=(10,3)) indicamos con el .subplots(1,2,) graficaruna fila con dos espacios para graficar.
## El axs[1].grid() añade una cuadricula al gráfico.

fig, axs = plt.subplots(1,2,figsize=(10,3))
axs[0].plot(datos_col['Año'],datos_col['Inmigrantes'])
axs[0].set_title('Inmigración de Colombianos hacia Canadá\n1980 - 2013')
axs[0].set_xlabel('Año')
axs[0].set_ylabel('Número de Inmigrantes')
axs[0].xaxis.set_major_locator(plt.MultipleLocator(5))
axs[0].grid()

axs[1].boxplot(datos_col['Inmigrantes'])
axs[1].set_title('Boxplot de la Inmigración de Colombianos\nhacia Canadá 1980 - 2013')
axs[1].set_xlabel('Colombia')
axs[1].set_ylabel('Número de Inmigrantes')
axs[1].grid()

plt.show()
```

```python
## Con .describe() entrega la estadistca descriptiva de los datos.

datos_col.describe()
```

```python
## Realizaremos una comparación entre varios paises.
## Con subplots(2,2) garantizamos tener 4 gráficos.
## El fig.subplots_adjust(hspace=0.5,wspace=0.3) ajusta el espacio entre los gráficos.
## Utilizamos ymin = 0, ymax = 7000 para establecer un valor mínimo y máximo entre todas las escalas de las gráficas.

fig, axs = plt.subplots(2,2,figsize=(10,6))
fig.subplots_adjust(hspace=0.5,wspace=0.3)
fig.suptitle('Inmigración de los 4 mayores países sudamericanos\nhacia Canadá desde 1980 hasta 2013')

axs[0,0].plot(df.loc['Colombia',anos])
axs[0,0].set_title('Colombia')

axs[0,1].plot(df.loc['Brasil',anos])
axs[0,1].set_title('Brasil')

axs[1,0].plot(df.loc['Argentina',anos])
axs[1,0].set_title('Argentina')

axs[1,1].plot(df.loc['Perú',anos])
axs[1,1].set_title('Perú')

ymin = 0
ymax = 7000

for ax in axs.ravel():
  ax.set_ylim(ymin,ymax)

for ax in axs.flat:
  ax.xaxis.set_major_locator(plt.MultipleLocator(5))
  ax.set_xlabel('Año')
  ax.set_ylabel('Número de Inmigrantes')
  ax.grid()

  fig.savefig('/content/drive/MyDrive/Colab Notebooks/Inmigrantes_Paises_Sudamerica.png',
            transparent=False,dpi=300,bbox_inches='tight')

plt.show()
```

```python
fig, ax = plt.subplots(figsize=(8,3))
ax.plot(datos_col['Año'],datos_col['Inmigrantes'],lw=2.5,color='g') #,marker='o')
ax.xaxis.set_major_locator(plt.MultipleLocator(5))
ax.set_title('Inmigración de colombianos hacia Canadá\n1980 - 2013', fontsize=18, loc='left')
ax.set_xlabel('Año',fontsize=14)
ax.set_ylabel('Número de Inmigrantes',fontsize=14)
ax.xaxis.set_tick_params(labelsize=12)
ax.yaxis.set_tick_params(labelsize=12)
ax.grid(linestyle='--')
plt.show()
```

```python
## Creamos un nuevo DataFrame llamado sudamerica.
## Utilizamos nuestro DataFrame original con el método query para consulutar sobre la región y América del Sur únicamente.

sudamerica = df.query('Region == "América del Sur"')
sudamerica.head(10)
```

```python
## Graficamos la cantida de migrantes para cada uno de estos paises.

colores = ['royalblue','orange','forestgreen','orchid',
           'purple','brown','slateblue','gray','olive',
           'navy','teal','tomato']

fig, ax = plt.subplots(figsize=(10,3))

ax.bar(sudamerica.index, sudamerica['Total'],color=colores)
ax.set_title('Inmigración de sudamericanos hacia Canadá\ndurante el periodo de 1980 hasta 2013',
             loc='left',fontsize=18)
ax.set_xlabel('')
ax.set_ylabel('Número de Inmigrantes', fontsize=14)
ax.xaxis.set_tick_params(labelsize=12)
ax.yaxis.set_tick_params(labelsize=12)

plt.show()
```

```python
## Corregimos el gráfico anterior donde se solapan los nombres de los paises.

colores = ['royalblue','orange','forestgreen','orchid',
           'purple','brown','slateblue','gray','olive',
           'navy','teal','tomato']

fig, ax = plt.subplots(figsize=(10,3))

ax.barh(sudamerica.index, sudamerica['Total'],color=colores)
ax.set_title('Inmigración de sudamericanos hacia Canadá\ndurante el periodo de 1980 hasta 2013',
             loc='left',fontsize=18)
ax.set_ylabel('')
ax.set_xlabel('Número de Inmigrantes', fontsize=14)
ax.xaxis.set_tick_params(labelsize=12)
ax.yaxis.set_tick_params(labelsize=12)

plt.show()
```

```python
## Organizamos con el método sort_values los valores del gráfico de menor a mayor.

sudamerica_sorted = sudamerica.sort_values('Total',ascending=True)

colores = ['royalblue','orange','forestgreen','orchid',
           'purple','brown','slateblue','gray','olive',
           'navy','teal','tomato']

fig, ax = plt.subplots(figsize=(10,3))

ax.barh(sudamerica_sorted.index, sudamerica_sorted['Total'],color=colores)
ax.set_title('Inmigración de sudamericanos hacia Canadá\ndurante el periodo de 1980 hasta 2013',
             loc='left',fontsize=18)
ax.set_ylabel('')
ax.set_xlabel('Número de Inmigrantes', fontsize=14)
ax.xaxis.set_tick_params(labelsize=12)
ax.yaxis.set_tick_params(labelsize=12)

plt.show()
```

```python
## Resaltamos nuestro país de estudio en este caso Colombia.

colores =[]
for pais in sudamerica_sorted.index:
  if pais == 'Colombia':
    colores.append('forestgreen')
  else:
    colores.append('silver')

fig, ax = plt.subplots(figsize=(10,3))

ax.set_frame_on(False)
ax.xaxis.set_visible(False)
ax.tick_params(axis='y',which='both',size=0)

ax.barh(sudamerica_sorted.index, sudamerica_sorted['Total'],color=colores)
ax.set_title('Sudamérica: Colombia es el segundo país con más migrantes\nhacia Canadá durante el periodo 1980 - 2013.',
             loc='left',fontsize=16)
ax.set_ylabel('')
ax.yaxis.set_tick_params(labelsize=10)
# ax.set_xlabel('Número de Inmigrantes', fontsize=13)
# ax.xaxis.set_tick_params(labelsize=12)

for i,j in enumerate(sudamerica_sorted['Total']):
  ax.text(j+20,i,str(j),color='black',fontsize=9,ha='left',va='center')

fig.savefig('/content/drive/MyDrive/Colab Notebooks/Inmigración_Sudamerica.png',
            transparent=False,dpi=300,bbox_inches='tight')

plt.show()
```

## Seaborn
https://seaborn.pydata.org/

```python
## Importamos la Biblioteca Seaborn.

import seaborn as sns
```

```python
## Configuramos un tema al dejarlo vacio () dejamos el tema por defecto.

sns.set_theme()
```

```python
## Creamos un nuevo DataFrame con los datos globales historicos de migración a Canada.

top_10 = df.sort_values('Total',ascending=False).head(10)
top_10
```

```python
## Creamos un gráfico de barras para el nuevo DataFrame.

ax = sns.barplot(data=top_10,x='Total',y=top_10.index,orient='h')
ax.set(title='Top 10 países con mayor migración hacia Canadá\n1980 - 2013',
       xlabel='Número de Inmigrantes',
       ylabel='Año')
plt.show()
```

```python
## Utilizamos las dos bibliotecas en conjunto.

fig, ax = plt.subplots(figsize=(8,4))

ax = sns.barplot(data=top_10,x='Total',y=top_10.index,orient='h')
ax.set_title('Top 10 países con mayor migración hacia Canadá\n1980 - 2013',
             loc='left',fontsize=16)
ax.set_xlabel('Número de Inmigrantes',fontsize=14)
ax.set_ylabel('')
plt.show()
```

```python
## Definimos una función que nos permita probar diferentes palestas de colores.

def generar_grafico(palette):
  fig, ax = plt.subplots(figsize=(8,4))
  ax.set_frame_on(False)
  ax.xaxis.set_visible(False)
  ax.tick_params(axis='y',which='both',size=0)
  ax = sns.barplot(data=top_10,x='Total',y= top_10.index,hue=top_10.index,orient='h',
                   palette=palette, legend=False)
  ax.set_title('Top 10 países con mayor migración hacia Canadá\n1980 - 2013',
              loc='left',fontsize=16)
  ax.set_xlabel('Número de Inmigrantes',fontsize=14)
  ax.set_ylabel('')
  for i,j in enumerate(top_10['Total']):
    ax.text(j+20,i,str(j),color='black',fontsize=9,ha='left',va='center')
  plt.show()
```

```python
def generar_grafico(palette):
  fig, ax = plt.subplots(figsize=(8,4))
  ax.set(xticklabels=[])
  ax = sns.barplot(data=top_10,x='Total',y= top_10.index,hue=top_10.index,orient='h',
                   palette=palette, legend=False)
  ax.set_title('Top 10 países con mayor migración hacia Canadá\n1980 - 2013',
              loc='left',fontsize=16)
  ax.set_xlabel('Número de Inmigrantes',fontsize=14)
  ax.set_ylabel('')
  sns.despine()
  for i,j in enumerate(top_10['Total']):
    ax.text(j+20,i,str(j),color='black',fontsize=9,ha='left',va='center')
  plt.show()
```

```python
import plotly.express as px
```

```python
## La Biblioteca Plotly nos ayuda a trabajar de manera mas interactiva nuestra gráfica.

fig = px.line(datos_col,x='Año',y='Inmigrantes',
              title='Inmigración de Colombianos hacia Canadá entre 1980 y 2013')
fig.update_traces(line_color='green',line_width=3)
fig.update_layout(width=800,height=400, xaxis={'tickangle':-45},
                  xaxis_title='Año',yaxis_title='Número de Inmigrantes',
                  font_family='Arial',font_size=14, font_color='grey')
fig.show()
```
