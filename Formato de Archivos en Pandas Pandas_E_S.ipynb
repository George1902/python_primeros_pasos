```python
from sqlalchemy import text
from sqlalchemy.exc import SQLAlchemyError

# Definimos la consulta
query = text("UPDATE clientes SET Grado_estudio = 'Nivel superior' WHERE ID_Cliente =5008808")

try:
    # Usar 'with' asegura que la conexión se cierre automáticamente, incluso si hay un error.
    # Esto evita el error "database is locked" en el futuro.
    with engine.connect() as connection:
        # Ejecutamos la consulta
        r_set = connection.execute(query)

        # OBLIGATORIO: En SQLAlchemy 2.0+ hay que confirmar los cambios
        connection.commit()

        print("# Registros Actualizados: ", r_set.rowcount)

except SQLAlchemyError as e:
    print(f"Error de SQLAlchemy: {e}")
except Exception as e:
    print(f"Ocurrió un error inesperado: {e}")
```
