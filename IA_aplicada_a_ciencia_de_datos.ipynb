### **2da Pregunta:** ¿Cuál fue la facturación por categoria em 2023?

**Observación:** Para la facturación, consideraremos aquí el producto entre el precio unitario y la cantidad, sumados al costo de envío de cada producto.


> #### <strong><font color="7b71ff" size=6>Prompt:</font></strong>

Tenemos un DataFrame llamado "df" y queremos construir un gráfico de barras horizontales que muestre la facturación por categoría de Zoop.

En "df", tenemos las siguientes columnas:
* categoria: categorías de los productos
* precio_unitario: precio de una unidad del producto
* cantidad: cantidad vendida del producto
* envio: costo de envío total de la compra.

Deseamos crear una visualización que muestre la facturación de cada categoría, donde la facturación se calcula multiplicando el precio unitario por la cantidad y sumando el costo de envío.

Utiliza las bibliotecas de visualización de datos de Python para crear esta visualización.


````python
# 1. Calcular la facturación para cada compra
df['facturacion'] = (df['precio_unitario'] * df['cantidad']) + df['envio']

# 2. Agrupar por categoría y sumar la facturación
facturacion_categoria = df.groupby('categoria')['facturacion'].sum().reset_index()

# 3. Ordenar los datos por facturación de mayor a menor
facturacion_categoria = facturacion_categoria.sort_values(by='facturacion', ascending=False)

# 4. Crear el gráfico de barras horizontales
plt.figure(figsize=(10, 6))
sns.barplot(x='facturacion', y='categoria', data=facturacion_categoria, palette='viridis',orient='h')

# 5. Añadir etiquetas y título
plt.title('Facturación por Categoría de Producto', fontsize=16)
plt.xlabel('Facturación (en MXN$)', fontsize=12)
plt.ylabel('Categoría', fontsize=12)

# 6. Mostrar el gráfico
plt.show()
````

### **3a Pregunta:** ¿Cómo fueron las ventas a lo largo de los meses en 2023?


> #### <strong><font color="7b71ff" size=6>Prompt:</font></strong>

Tenemos un DataFrame llamado "df" y queremos construir un gráfico de líneas que muestre las ventas totales mensuales de Zoop.

En "df", tenemos las siguientes columnas:
* fecha: con las fechas de las compras en formato datetime (`aaaa-mm-dd`)
* facturación: calculada como el producto del precio unitario y la cantidad, sumado al costo de envío de cada venta.

Deseamos crear una visualización que muestre las ventas por mes. Primero, agrupa los datos por mes y luego crea una nueva columna llamada "mes" que contenga el nombre de cada mes, traduciendo los meses del índice mediante el siguiente diccionario. Utiliza la columna "fecha" para agrupar los datos y construir el gráfico. La columna "mes" solo debe usarse para cambiar las etiquetas del eje x.

```
meses = {
    'January': 'Ene',
    'February': 'Feb',
    'March': 'Mar',
    'April': 'Abr',
    'May': 'May',
    'June': 'Jun',
    'July': 'Jul',
    'August': 'Ago',
    'September': 'Sep',
    'October': 'Oct',
    'November': 'Nov',
    'December': 'Dic'
}
```


Utiliza las bibliotecas de visualización de datos de Python para crear esta visualización.


````python
# Supongamos que ya se ha calculado la columna 'facturacion'

# 1. Crear una nueva columna 'mes' que extrae el mes de la columna 'fecha'
df['mes'] = df['fecha'].dt.month_name()

# 2. Agrupar los datos por mes y calcular la facturación total mensual
ventas_mensuales = df.groupby(df['fecha'].dt.to_period('M'))['facturacion'].sum().reset_index()

# 3. Convertir el periodo de fecha a tipo datetime para mantener formato de fecha
ventas_mensuales['fecha'] = ventas_mensuales['fecha'].dt.to_timestamp()

# 4. Crear una columna de nombre del mes en español usando el diccionario proporcionado
meses = {
    'January': 'Ene',
    'February': 'Feb',
    'March': 'Mar',
    'April': 'Abr',
    'May': 'May',
    'June': 'Jun',
    'July': 'Jul',
    'August': 'Ago',
    'September': 'Sep',
    'October': 'Oct',
    'November': 'Nov',
    'December': 'Dic'
}

ventas_mensuales['mes'] = ventas_mensuales['fecha'].dt.month_name().map(meses)

# 5. Crear el gráfico de líneas de ventas mensuales
plt.figure(figsize=(10, 6))
sns.lineplot(x='mes', y='facturacion', data=ventas_mensuales, marker='o', sort=False)

# 6. Añadir etiquetas y título
plt.title('Ventas Totales Mensuales de Zoop', fontsize=16)
plt.xlabel('Mes', fontsize=12)
plt.ylabel('Facturación Total', fontsize=12)

# 7. Mostrar el gráfico
plt.show()
````

---

Cuando tenemos una cantidad mayor de períodos en nuestros datos, vale la pena recurrir a gráficos de líneas, que son ideales para series temporales. Dado que nuestros datos son mensuales, utilizar un gráfico de columnas, por ejemplo, reduciría la percepción de la secuencia de los datos. Las líneas ofrecen una mayor sensación de continuidad.

````python
# 1. Crear una nueva columna 'trimestre' que extrae el trimestre de la columna 'fecha'
df['trimestre'] = df['fecha'].dt.to_period('Q')

# 2. Agrupar los datos por trimestre y método de pago, sumando la facturación
ventas_trimestre = df.groupby(['trimestre', 'metodo_pago'])['facturacion'].sum().unstack()

# 3. Crear el gráfico de barras apiladas
ventas_trimestre.plot(kind='bar', stacked=True, figsize=(10, 6), colormap='viridis')

# 4. Añadir etiquetas y título
plt.title('Ventas por Trimestre y Método de Pago', fontsize=16)
plt.xlabel('Trimestre', fontsize=12)
plt.ylabel('Facturación Total', fontsize=12)
plt.xticks(rotation=45)  # Rotar las etiquetas del eje x para mayor claridad

# 5. Mostrar el gráfico
plt.legend(title='Método de Pago por trimestre de 2023')
plt.tight_layout()
plt.show()
````

## **4.4 - Agregar elementos visuales**

---


¿Qué tal el uso de textos informativos, verdad? ¿Y si ahora añadimos otros elementos visuales a uno de nuestros gráficos?

Vamos a elegir ahora el gráfico de líneas (Ventas totales mensuales de Zoop) y aplicar una técnica más a él mediante el uso de la IA.

````python
import matplotlib.pyplot as plt
from matplotlib.ticker import FuncFormatter

def ventas_trimestre():
    # 1. Crear la columna trimestre
    df["trimestre"] = df["fecha"].dt.to_period("Q")

    # 2. Agrupar por trimestre y método de pago
    ventas_trimestre = (
        df.groupby(["trimestre", "metodo_pago"])["facturacion"]
          .sum()
          .unstack()
    )

    # 3. Crear figura y ejes
    fig, ax = plt.subplots(figsize=(10, 6))
    fig.patch.set_facecolor(GRIS_1)
    ax.set_facecolor(GRIS_1)

    # 4. Gráfico de barras apiladas con identidad Zoop
    ventas_trimestre.plot(
        kind="bar",
        stacked=True,
        ax=ax,
        color=[BLANCO, GRIS_3, GRIS_1, GRIS_3]
    )

    # 5. Título
    ax.set_title(
        "Ventas por Trimestre y Método de Pago en 2023",
        fontsize=18,
        color=GRIS_5
    )

    # 6. Configuración del eje X
    ax.set_xlabel("")
    ax.set_xticklabels(ax.get_xticklabels(), rotation=0, fontsize=12, color=AZUL_1)

    # 7. Configuración del eje Y
    ax.set_ylabel("")
    ax.yaxis.set_major_formatter(
        FuncFormatter(lambda x, pos: f"MXN$ {int(x/1e6)} M")
    )

    # 8. Quitar bordes del gráfico
    for spine in ["top", "bottom", "left", "right"]:
        ax.spines[spine].set_visible(False)

    # 9. Cuadrícula punteada en eje Y
    ax.grid(axis="y", linestyle=":", alpha=0.6)

    # 10. Leyenda
    ax.legend(title="Método de Pago por trimestre de 2023")

    plt.tight_layout()
    return plt
````

````python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
from matplotlib.ticker import FuncFormatter

# Assuming color variables are defined in the global scope from previous cells
# ROJO_1, ROJO_2, ROJO_3, AZUL_1, AZUL_2, AZUL_3, BLANCO, GRIS_1, GRIS_2, GRIS_3, GRIS_4, GRIS_5, AQUA_1, AQUA_2, AQUA_3
# Assuming df and meses are also globally available

# =========================
# Funciones de los gráficos
# (cada una debe crear y devolver la figura)
# =========================

def metodos_de_pago():
    metodos_de_pago = df['metodo_pago'].value_counts()
    total_clientes = metodos_de_pago.sum()
    porcentaje = (metodos_de_pago[0] + metodos_de_pago[1]) / total_clientes * 100

    fig, ax = plt.subplots(figsize=(10, 6), facecolor=GRIS_1)
    ax.set_facecolor(GRIS_1)
    ax.bar(metodos_de_pago.index, metodos_de_pago, color=ROJO_1)

    for i, v in enumerate(metodos_de_pago):
        ax.text(i, v + 0.1, str(v), ha='center', va='bottom', color=AZUL_1, fontsize=12)

    ax.text(1.5, 2500, f"$\\bf{porcentaje:.2f}$% de los clientes utilizan Tarjeta de Crédito o Transferencia \npara el pago. Sería interesante considerar asociaciones con \nbancos para la creación de un Zoop Pay con el fin de fidelizar \na nuestra clientela.",
            ha='left', va='center', color=AZUL_3, fontsize=10)

    ax.set_title('Métodos de Pago más Utilizados en 2023', fontsize=18, color=GRIS_5)
    ax.tick_params(axis='x', labelsize=12, colors=AZUL_1)
    ax.set_yticks([])
    ax.set_xlabel('Método de Pago', fontsize=12, color=AZUL_1)
    ax.spines[['top', 'right', 'left']].set_visible(False)
    ax.grid(False)

    return fig


def ventas_mensuales():
    df_copy = df.copy()
    df_copy['fecha'] = pd.to_datetime(df_copy['fecha'])
    df_copy['facturacion'] = df_copy['precio_unitario'] * df_copy['cantidad'] + df_copy['envio']
    df_copy['mes'] = df_copy['fecha'].dt.strftime('%B')
    ventas_mensuales_df = df_copy.groupby('mes')['facturacion'].sum().reindex(meses.keys())

    fig, ax = plt.subplots(figsize=(12, 6))
    fig.patch.set_facecolor(GRIS_1)
    ax.set_facecolor(GRIS_1)

    ax.plot(ventas_mensuales_df.index, ventas_mensuales_df, marker='o', color=AZUL_1, linewidth=2)

    ax.set_title('Ventas totales mensuales en 2023', fontsize=18, color=GRIS_5)

    ax.set_xticks(ventas_mensuales_df.index)
    ax.set_xticklabels([meses[mes] for mes in ventas_mensuales_df.index], fontsize=12, color=AZUL_1)

    ax.set_ylabel('')
    ax.set_yticklabels([f'MXN$ {x/1e6 :,.2f} M' for x in ax.get_yticks()], fontsize=12, color=AZUL_1)

    ax.set_xlabel('')

    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)

    for i, point in enumerate([(4, ventas_mensuales_df.iloc[4]), (7, ventas_mensuales_df.iloc[7]), (10, ventas_mensuales_df.iloc[10]), (11, ventas_mensuales_df.iloc[11])], start=1):
        ax.scatter(point[0], point[1], s=200, color='none', edgecolors=ROJO_1, alpha=0.7)
        ax.text(point[0] - 0.3, point[1] + 0.1, str(i), fontsize=12, color=ROJO_1)

    return fig


def facturacion_categoria():
    df_copy = df.copy()
    df_copy["facturacion"] = (df_copy["precio_unitario"] * df_copy["cantidad"]) + df_copy["envio"]

    facturacion_categoria_df = (
        df_copy.groupby("categoria")["facturacion"]
          .sum()
          .reset_index()
          .sort_values(by="facturacion", ascending=False)
    )
    porcentaje = (
        facturacion_categoria_df.iloc[0]["facturacion"]
        / facturacion_categoria_df["facturacion"].sum()
    ) * 100

    fig, ax = plt.subplots(figsize=(10, 6))
    fig.patch.set_facecolor(GRIS_1)
    ax.set_facecolor(GRIS_1)

    sns.barplot(
        x="facturacion",
        y="categoria",
        data=facturacion_categoria_df,
        color=AQUA_1,
        orient="h",
        ax=ax
    )

    ax.set_title(
        "Facturación por categoría de producto en 2023",
        fontsize=18,
        color=GRIS_5
    )

    ax.set_xlabel("")
    ax.set_ylabel("")
    ax.set_xticks([])

    ax.tick_params(axis="y", labelsize=12, colors=AZUL_1)

    for spine in ax.spines.values():
        spine.set_visible(False)

    for i, valor in enumerate(facturacion_categoria_df["facturacion"]):
        ax.text(
            valor,
            i,
            f"{valor:,.0f}",
            va="center",
            ha="left",
            fontsize=12,
            color=AZUL_1
        )

    ax.text(
        0.4e7,
        5,
        f"$\\bf{porcentaje:.2f}$% de las ventas corresponden únicamente a las ventas de Electrónica.\n" +
        "Esto equivale a casi el doble en relación con todas las demás categorías\n" +
        "presentes en el portafolio de Zoop.",
        fontsize=12,
        color=AZUL_3
    )

    return fig


def ventas_trimestre():
    df_copy = df.copy()
    df_copy["trimestre"] = df_copy["fecha"].dt.to_period("Q")
    ventas_trimestre_df = (
        df_copy.groupby(["trimestre", "metodo_pago"])["facturacion"]
          .sum()
          .unstack()
    )
    fig, ax = plt.subplots(figsize=(10, 6))
    fig.patch.set_facecolor(GRIS_1)
    ax.set_facecolor(GRIS_1)

    colors_palette = [BLANCO, GRIS_3, GRIS_1, GRIS_3] # Following prompt, though GRIS_1 matches background

    ventas_trimestre_df.plot(
        kind="bar",
        stacked=True,
        ax=ax,
        color=colors_palette
    )

    ax.set_title(
        "Ventas por Trimestre y Método de Pago en 2023",
        fontsize=18,
        color=GRIS_5
    )

    ax.set_xlabel("")
    ax.set_xticklabels([f"Q{t.quarter} {t.year}" for t in ventas_trimestre_df.index], rotation=0, fontsize=12, color=AZUL_1)

    ax.set_ylabel("")
    ax.yaxis.set_major_formatter(
        FuncFormatter(lambda x, pos: f"MXN$ {int(x/1e6)} M")
    )

    for spine in ["top", "bottom", "left", "right"]:
        ax.spines[spine].set_visible(False)

    ax.grid(axis="y", linestyle=":", alpha=0.6)

    ax.legend(title="Método de Pago por trimestre de 2023")

    fig.tight_layout()
    return fig


def proporcion_cashback():
    cashback_proporcion = df['cashback'].value_counts()

    fig, ax = plt.subplots(figsize=(6, 6))
    fig.patch.set_facecolor(GRIS_1)
    ax.set_facecolor(GRIS_1)

    wedges, texts, autotexts = ax.pie(
        cashback_proporcion,
        labels=cashback_proporcion.index,
        colors=[GRIS_3, AZUL_1],   # No, Sí
        autopct="%1.1f%%",
        startangle=180,
        wedgeprops={"width": 0.4},
        pctdistance=0.8
    )

    ax.set_title(
        "¿Participan en el programa de cashback?",
        fontsize=18,
        color=GRIS_5
    )

    for text_obj, wedge_obj in zip(texts, wedges):
        text_obj.set_color(wedge_obj.get_facecolor())
        text_obj.set_fontsize(12)

    for autotext_obj in autotexts:
        autotext_obj.set_color(BLANCO)
        autotext_obj.set_fontsize(12)

    ax.axis("equal")

    return fig


def distribucion_calificaciones():
    fig, ax = plt.subplots(figsize=(10, 6))
    fig.patch.set_facecolor(GRIS_1)
    ax.set_facecolor(GRIS_1)

    sns.histplot(
        df["nota"],
        bins=11,
        color=ROJO_1,
        kde=False,
        ax=ax
    )

    ax.set_title(
        "Distribución de las evaluaciones de las compras en 2023",
        fontsize=18,
        color=GRIS_5
    )

    ax.set_xlabel("Calificación (0-10)", fontsize=12, color=AZUL_1)
    ax.set_ylabel("Frecuencia", fontsize=12, color=AZUL_1)

    ax.set_xticks(range(0, 11))
    ax.tick_params(axis="x", labelsize=12, colors=AZUL_1)
    ax.tick_params(axis="y", labelsize=12, colors=AZUL_1)

    for spine in ["top", "bottom", "left", "right"]:
        ax.spines[spine].set_visible(False)

    ax.set_axisbelow(True)
    ax.grid(axis="y", linestyle="--", alpha=0.5)

    texto_anotacion = (
        f"Promedio de calificaciones = {df['nota'].mean().round(2)}\n" +
        f"Calificación más común = {df['nota'].value_counts().idxmax()}\n" +
        f"Frecuencia más común = {df['nota'].value_counts().max()}"
    )

    ax.annotate(
        texto_anotacion,
        xy=(8.7, 2250),
        xytext=(2, 1800),
        fontsize=12,
        color=AZUL_1,
        bbox=dict(facecolor=BLANCO, edgecolor=GRIS_3),
        arrowprops=dict(arrowstyle="->", color=AZUL_1)
    )

    return fig


def box_plot_edad_sexo():
    stats = df.groupby("sexo_biologico")["edad"]
    min_edad = stats.min()
    q1_edad = stats.quantile(0.25)
    mediana_edad = stats.median()
    q3_edad = stats.quantile(0.75)
    max_edad = stats.max()

    fig, ax = plt.subplots(figsize=(10, 6))
    fig.patch.set_facecolor(GRIS_1)
    ax.set_facecolor(GRIS_1)

    sns.boxplot(
        x="sexo_biologico",
        y="edad",
        data=df,
        palette=[AZUL_1, AQUA_1],
        ax=ax
    )

    ax.set_title(
        "Distribución de la Edad por Sexo Biológico de los clientes",
        fontsize=18,
        color=GRIS_5
    )

    ax.set_xlabel("")
    ax.set_ylabel("Edad", fontsize=12, color=AZUL_1)
    ax.tick_params(axis="x", labelsize=12, colors=AZUL_1)
    ax.tick_params(axis="y", labelsize=12, colors=AZUL_1)

    for spine in ["top", "bottom", "left", "right"]:
        ax.spines[spine].set_visible(False)

    ax.set_axisbelow(True)
    ax.grid(axis="y", linestyle="--", alpha=0.5)

    texto_femenino = (
        f"Mínimo = {min_edad.loc['Femenino']}\n" +
        f"1º quartil = {q1_edad.loc['Femenino']}\n" +
        f"2º quartil = {mediana_edad.loc['Femenino']}\n" +
        f"3º quartil = {q3_edad.loc['Femenino']}\n" +
        f"Máximo = {max_edad.loc['Femenino']}"
    )

    ax.annotate(
        texto_femenino,
        xy=(-0.1, 45),
        xytext=(-0.4, 50),
        fontsize=12,
        color=AZUL_1,
        bbox=dict(facecolor=BLANCO, edgecolor=GRIS_3),
        arrowprops=dict(arrowstyle="->", color=AZUL_1)
    )

    texto_masculino = (
        f"Mínimo = {min_edad.loc['Masculino']}\n" +
        f"1º quartil = {q1_edad.loc['Masculino']}\n" +
        f"2º quartil = {mediana_edad.loc['Masculino']}\n" +
        f"3º quartil = {q3_edad.loc['Masculino']}\n" +
        f"Máximo = {max_edad.loc['Masculino']}"
    )

    ax.annotate(
        texto_masculino,
        xy=(0.9, 45),
        xytext=(0.6, 50),
        fontsize=12,
        color=AZUL_1,
        bbox=dict(facecolor=BLANCO, edgecolor=GRIS_3),
        arrowprops=dict(arrowstyle="->", color=AZUL_1)
    )

    return fig


# =========================
# Guardar las visualizaciones
# =========================

grafico_metodos_de_pago = metodos_de_pago()
grafico_metodos_de_pago.savefig("grafico_metodos_de_pago.png", bbox_inches="tight")
plt.close(grafico_metodos_de_pago)

grafico_ventas_mensuales = ventas_mensuales()
grafico_ventas_mensuales.savefig("grafico_ventas_mensuales.png", bbox_inches="tight")
plt.close(grafico_ventas_mensuales)

grafico_facturacion_categoria = facturacion_categoria()
grafico_facturacion_categoria.savefig(
    "grafico_facturacion_categoria.png", bbox_inches="tight"
)
plt.close(grafico_facturacion_categoria)

grafico_ventas_trimestre = ventas_trimestre()
grafico_ventas_trimestre.savefig(
    "grafico_ventas_trimestre.png", bbox_inches="tight"
)
plt.close(grafico_ventas_trimestre)

grafico_proporcion_cashback = proporcion_cashback()
grafico_proporcion_cashback.savefig(
    "grafico_proporcion_cashback.png", bbox_inches="tight"
)
plt.close(grafico_proporcion_cashback)

grafico_distribucion_calificaciones = distribucion_calificaciones()
grafico_distribucion_calificaciones.savefig(
    "grafico_distribucion_calificaciones.png", bbox_inches="tight"
)
plt.close(grafico_distribucion_calificaciones)

grafico_box_plot_edad_sexo = box_plot_edad_sexo()
grafico_box_plot_edad_sexo.savefig(
    "grafico_box_plot_edad_sexo.png", bbox_inches="tight"
)
plt.close(grafico_box_plot_edad_sexo)
````
